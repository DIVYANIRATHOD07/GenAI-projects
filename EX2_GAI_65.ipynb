```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
```

```python
# Arithmatic

print("Addition =", a + b)
print("Subtraction =", a - b)
print("Multiplication =", a * b)
print("Division =", a / b)
print("Floor Division =", a // b)
print("Modulus =", a % b)
print("Exponent =", a ** b)
```

```python
# Assignment Operations
x = a
print("Initial x =", x)

x += b
print("x += b :", x)

x -= b
print("x -= b :", x)

x *= b
print("x *= b :", x)

x //= b
print("x //= b :", x)

x %= b
print("x %= b :", x)
```

```python
# Logical Operations
print("a > 0 and b > 0 =", a > 0 and b > 0)
print("a > 0 or b > 0 =", a > 0 or b > 0)
print("not(a > b) =", not(a > b))
```

```python
print("Bitwise AND (a & b) =", a & b)
```

```python
# comparison operation
print("a == b :", a == b)
print("a != b :", a != b)
print("a > b :", a > b)
print("a < b :", a < b)
print("a >= b :", a >= b)
print("a <= b :", a <= b)
```

```python
import tensorflow as tf
import numpy as np
print("TensorFlow Version:", tf.__version__)
```

```python
# Scalar (0-D)
scalar = tf.constant(10)

# Vector (1-D)
vector = tf.constant([1, 2, 3, 4])

# Matrix (2-D)
matrix = tf.constant([[1, 2], [3, 4]])

# 3-D Tensor
tensor_3d = tf.constant([[[1, 2], [3, 4]],
                         [[5, 6], [7, 8]]])

print("Scalar:\n", scalar)
print("Vector:\n", vector)
print("Matrix:\n", matrix)
print("3-D Tensor:\n", tensor_3d)
```

```python
print("Shape:", matrix.shape)
print("Rank:", tf.rank(matrix))
print("Data Type:", matrix.dtype)
```

```python
a = tf.constant([10, 20, 30])
b = tf.constant([2, 4, 5])

print("Addition:", tf.add(a, b))
print("Subtraction:", tf.subtract(a, b))
print("Multiplication:", tf.multiply(a, b))
print("Division:", tf.divide(a, b))
```

```python
x = tf.constant([1, 4, 9, 16], dtype=tf.float32)

print("Square:", tf.square(x))
print("Square Root:", tf.sqrt(x))
print("Mean:", tf.reduce_mean(x))
print("Maximum:", tf.reduce_max(x))
```

```python
x = tf.constant([1, 2, 3, 4, 5, 6])

reshaped = tf.reshape(x, (2, 3))

print("Original Tensor:", x)
print("Reshaped Tensor:\n", reshaped)
```

```python
tensor = tf.constant([10, 20, 30, 40])

numpy_array = tensor.numpy()

print("TensorFlow Tensor:", tensor)
print("NumPy Array:", numpy_array)
```
